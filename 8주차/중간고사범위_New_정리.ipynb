## 0. 인공지능이란

인공지능이란 컴퓨터가 인간의 지능적 능력들을 모방하거나 구현하여 문제를 해결하도록 하는 기술이다.  

## 1. 인공지능에서 지능이란

= 인공지능에서 “지능”은 환경을 인식하고 경험을 바탕으로 학습하고 지식을 통해 추론과 문제 해결을하며 필요 시 창의성을 가지고 지식을 확장하는 능력들이 합쳐진 개념입니다.  

- 지각 : 새로운 외부 정보를 인식하는 것
- 학습 : 데이터를 기반으로 경험한 내용을 통해 지식을 습득하고 성능을 높여가는 것
- 추론 : 기존의 주어진 정보로부터 새로운 결론을 도출하는 것.
- 문제 해결 : 정해진 목표를 위해서 전략을 세우고 실행하는 능력
- 의사결정 : 여러가지 선택지가 있을 때 특정 기준을 바탕으로 선택하는 과정 또는 능력    (이건 faq에는 없음.)
- 적응 : 새로운 상황이나 환경 변화에 맞춰서 지식을 수정하거나 새롭게 정리하고 행동을 조절하는 능력
- 창의성 : 기존의 지식과 아이디어를 합쳐서 새로운 아이디어나 해결법을 만드는 능력
- 사회적 지능 :  상대방 (인간 또는 기계)를 이해하고 상황에 맞는 적절한 반응을 보이며 목표 달성을 위해 협력하는 능력

## 2. 인공지능의 학습 종류 3가지

= 인공지능의 학습 종류에는 지도학습, 비지도학습, 강화학습이 있다.

- 지도학습 : 입력데이터와 그에 대응하는 정답 데이터가 모델에게 함께 주어진다. 모델은 입력을 받아서 출력을 만들어내는 함수(학습된 규칙)인데,  모델이 입력과 출력의 관계를 학습하는 것을 지도학습이라고 한다.
    - 예시: 사진을 주고 고양이다. 환자 데이터를 주고 혈압 수치 주기
    - 특징 : 데이터에 정답이 있다.
    - 장단점 : 정확한 예측이 가능하고 틀리면 수정이 바로 가능하기에 학습이 안정적이다./대량의 라벨링(정답이 달림) 데이터가 필요해서 라벨링 작업을 하는 사람이 필요하고 데이터를 분류하고 검수하는 시간이 필요한 등의 비용이 많이 든다.
- 반지도학습 : 일부 입력 데이터에만 정답이 있고, 나머지는 정답이 없는 데이터를 함께 활용하여 모델이 학습하는 방식이다.
    - 예시 : 시험지 100장 중 10장만 채점하고 나머지 90장은 정답 없음. 10장을 바탕으로 나머지 데이터 구조를 더 잘 이해하도록 규칙을 확장하여 학습
    - 특징 : 일부만 정답이 있다. 실제 산업/의료 분야에서 많이 활용된다.
    - 장단점 : 소량의 라벨로도 성능을 높일 수 있고 라벨링 비용이 줄어든다./잘못된 라벨이 있거나 데이터 종류 비율이 한쪽으로 쏠리면 모델이 편향된 학습을 하며 성능이 떨어질 수 있다.
- 강화학습 : 인공지능이 어떤 규칙과 상황의 환경 속에서 선택 가능한 행동을 했을 때 행동의 결과로 좋음, 나쁨 등의 점수 즉 보상을 받아 누적 보상을 최대화 하는 것을 목적으로 학습하는 방식이다.
    - 예시 : 체스 바둑 AI(당장에 이기는 수 x 나중에 이기는 전략o)
    - 특징 : 정답이 정해져 있지 않음. 시행착오를 통해 보상을 기반으로 학습 함
    - 장단점 : 자율적으로 의사결정이 가능하고 복잡한 문제 해결 가능/학습 속도가 느리고 보상을 언제 얼마나 줄지 설계 어려움
    

-faz에는 없지만!

- 비지도학습 : 정답이 없이 입력 데이터만을 이용해 데이터의 구조나 패턴을 스스로 찾는 학습 방식이다.
    - 예시 : 군집화-고객 데이터를 비슷한 데이터까리 그룹화 하거나 구조를 찾음
    - 특징 : 정답도 보상도 없음. 데이터의 구조를 분석
    - 장단점 : 라벨링 비용 필요 없음, 대규모 데이터 활용 가능, 숨겨진 패턴 발견 가능/왜 이렇게 묶인 건지 결과 해석이 어려울 수 있음, 정확한 정답 기준이 없기 때문에 성능 평가도 어려움
- 강화 학습 vs 비지도 학습
    - 목표가 다름. 어떻게 행동해야 보상이 최대일까?  vs  이 데이터는 어떤 구조인가?
    - 행동 학습 vs 구조 분석
    - 행동하면서 점수 받기 vs 데이터 이해하기.      (강화학습은 데이터 이해 없이 그냥 행동!하는 거!)

## 3. 전통적인 프로그래밍 방법과 인공지능 프로그램의 차이점

= 두 가지 방식은 문제를 해결하는 접근 방식에서 차이가 있다.

- 전통적인 프로그래밍 방법
    - 규칙 기반 : 사람이 문제 해결 절차와 규칙을 직접 정의한다.
    - 입력 + 규칙 ⇒ 출력    구조
    - 장점 : 예측 가능성, 디버깅이 용이, 결과가 항상 동일
    - 단점 : 규칙이 복잡하거나, 불명확한 문제(자연어 처리, 얼굴 인식)는 사람이 규칙을 정의하기 어려움
- 인공지능 프로그램
    - 데이터 기반 학습 : 데이터를 통해 컴퓨터가 모델을 스스로 학습힌다.
    - 입력 + 출력(정답 데이터) ⇒ 규칙 형성
    - 장점 : 복잡한 것들, 사람이 정의하기 어려운 문제도 해결 가능, 새로운 패턴에도 적용 가능
    - 단점 : 학습 데이터 품질에 따라 성능이 좌우됨. 내부 작동 원리가 불투명할 수 있음(블랙박스)

## 4. 딥러닝과 머신러닝의 차이점

 딥러닝과 머신러닝의 특징 추출 관점에서의 차이점

- 머신러닝
    - 사람이 정의 : 도메인 지식을 활용하여 특징을 설계함
        - ex) 이미지에서 모서리, 색비율 등 중요한 정보만 보는 것, 심장 신호 그래프에서 상승 시간, 피크 간격 등 핵심 숫자만 보는 것. 전체가 아닌 특정 요소만 정해서 보는 것
    - 모델은 사람이 설계한 특징을 입력으로 받아 학습한다.
    - 따라서 모델의 성능이 데이터 자체보다 사람이 설계한 특징에 크게 의존한.  AI에게 어떤 정보를 주느냐의 수준이 성능을 크게 좌우한다.
- 딥러닝
    - 스스로(자동) 특징 학습 : 원시 데이터에서 직접 중요한 정보(특징)를 학습한다.
    - 신경망이 몇단계로 구성되어있는지, 그 계층 깊이에 따
        - 하위 계층 : 저수준 특징 학습
        - 상위 계층 : 고수준 특징 학습
        
        선 → 모서리 → 형태 → 얼굴 → 고양이(저수준 → 고수준)
        
    - 특징 설계(데이터에서 무엇을 뽑을지)에 사람 개입이 거의 필요 없고 데이터만 충분하면 복잡한 특징을 스스로 추출한다.

## 5. Classification과 Regression의 주된 차이점은?

= 핵심 차이는 결과가 이산적이냐 연속적이냐에  있다.  이산적이다는 라벨, 범주로 나뉘는 것이고 연속적이다는 값의 범위라는 것이다.

= Classification은 데이터를 그룹으로 분류하여(범주로) 구분하는 문제이고, Regression은 연속적인 값을 예측하는 문제이다.

- 분류 Classification
    - 출력값 : 이산적
    - 의미 : 입력 데이터가 어느 카테고리에 속하는지 판별함
    - 예시 : 스트레스 상태가 정상, 경도, 중증 중 어느 상태인지, 환자의 질병이 있는 지 없는 지
- 회귀 Regression
    - 출력값 : 연속적
    - 의미 : 입력 데이터를 통해 숫자의 크기나 양을 예측
    - 예시 : 스트레스에 따른 혈압 수치를 예측,  온도 예측

회귀란?

입력데이터를 기반으로하여 연속적인 수치 값을 예측하는 문제를 말한다.

분류란?

입력데이터를 기반으로하여 데이터를 여러 카테고리 종류 중 하나로 구분하는 문제를 말한다.

2. 데이터가 충분하지 않으면 해가 불안정해진다.
    1. 파라미터가 많아짐에 따라 그걸 구할 데이터 개수가 많아야 한다. 그러나 데이터가 부족하면 여러가지 해가 존재하거나 특정 방향으로만 잘 알고 새로운 데이터에는 모르고 안 맞는 과적합이 되어버림
3. 직관전 어려움
    1. 2차원에서 직선 하나를 찾는 것이랑 3차원에서 평면을 찾는 문제는 난이도가 다르다.  

## 7. 규제와 스케일링의 차이점은? 각각의 목적은?   30%

- 스케일링 : 서로 다른 크기의 데이터와 데이터의 단위 차이를 비슷한 범위로 맞추어 모델이 공정하게 학습하도록 하는 과정
    - 방법
        - 전처리 단계에서 표준화, 정규화 하는 등 데이터의 크기/범위를 맞추어 균형 있는 어느 한쪽의 정보만 과하게 중요해지지 않게끔 학습
        - 모델 내부에서 학습 중간 표현을 정규화하여 기울기 폭주(고칠 크기가 너무 커짐) 소실(너무 작아짐) 방지, 안정적 수렴
    - 효과 : 최적값에 더 빨리 도달하게 함(경사하강법 수렴 가속), 거리 기반 모델의 공정성 확보
- 규제 : 모델이 훈련 데이터에 과도하게 적합하지 않도록 복잡도를 제어하고 일반화 성능 확보
    - 방법
        - 전처리 단계에서 Data Augmentation(데이터에 변형 추가하여 모델이 둔하게 함) Feature Selection(단순하고 의미 있는 피처만 사용함)
        - 모델 내부에서 L1 규제(일부 가중치를 0으로)와 L2규제(모든 가중치를 줄이기)를 함.

| **구분** | **스케일링** | **규제** |
| --- | --- | --- |
| **전처리 단계** | 데이터 크기/범위 변환 (표준화, 정규화) → 공정한 입력 | 데이터/피처 제약 (Augmentation, Feature Selection, PCA) → 단순화 |
| **모델 단계** | BatchNorm, LayerNorm 등으로 내부 표현 안정화 | L1/L2 패널티, Dropout 등으로 파라미터 제약 → 과적합 방지 |
| **핵심 목적** | 수치적 안정성 & 효율적 최적화 | 복잡도 제어 & 일반화 성능 확보 |
| **피처 관점** | 모든 피처를 같은 기준선에 세움 | 덜 중요한 피처 제거(L1) 또는 중요도 축소(L2) |

## 8. Overfitting vs UnderFitting 목적과 차이점은?

- 과적합 Overfitting
    - 정의 : 모델이 훈련 데이터에 지나치게 알맞아서, 훈련 데이터에만 성능이 높고 새로운 테스트 데이터에서는 성능이 떨어지는 현상
    - 원인 : 모델이 너무 복잡하거나, 파라미터가 많아서 데이터의 의미없는 잡음까지 학습해버림
    - 목적 : (실패)새로운 데이터에서도 잘 맞추는 능력이 떨어짐. 데이터 패턴을 제대로 이해한 것이 아니라 우연한 특성을 외워버린 것이기 때문
    - 특징 ; 훈련 오차가 적음, 테스트 오차가 큼, 복잡한 모델은 더 많이 훈련 데이터를 외우기 좋아서 이러한 모델에서 자주 발생.
- 과소적합 Underfitting
    - 정의 : 모델이 너무 단순하여 데이터의 중요한 패턴조차 제대로 학습하지 못하는 현상
    - 원인 : 모델의 복잡도가 너무 낮거나 학습이 충분히 이루어지지 않음
    - 목적 : (실패)데이터 표현이 부족, 중요한 구조조차 잡아내지를 못하게 됨.
    - 특징 : 훈련 오차가 높음, 테스트 오차도 높음, 단순한 모델에서 자주 발생

## 10. 전처리(Preprocessing)의 목적과 방법은?  (노이즈, 이상치, 결측치)

= 받은 시험지를 풀이용으로 정리하는 과정 (정리 영역)

- 전처리의 목적
    1. 데이터 품질 향상 : 잘못된 값이나 잡음이나 불완전한 데이터를 정리하여 신뢰성 있는 입력 확보
    2. 모델 성능 개선 : 불필요한 변동을 줄이고 중요한 패턴이 더 잘 드러나도록 함
    3. 일반화 성능 확보 : 데이터의 잡음 때문에 생기는 Overfitting 위험을 줄임
    4. 계산 효율성 향상 : 불필요한 이상치나 결측치를 줄여 학습 속도와 안정성 확보
- 전처리방법
    1. 노이즈 처리
        1. 평탈화(그래프를 매끄럽게 만듬), 이동평균(주변 값 평균 내서 부드럽게 만듬)
        2. 변환 : 로그 변환, 정규화로 극단값 완화
        3. 신호 처리 : 잡음을 제거하며 필터링
    2. 이상치 처리 : 정상 범위를 넘어선 값이 모델을 왜곡하지 않도록 조정
        1.   탐지 : 통계 기반, 시각화
        2. 처리 : 제거, 중앙값으로 대체, 스케일링하여 변환
    3. 결측치 처리 : 데이터 누락으로 인한 정보 손실을 최소화
        1. 단순 대체(평균값) 하거나 KNN, MICE, Regression등으로 고급 대체
        2. 제거 : 결측치가 너무 많은 행/열은 삭제
        3. 예측 기반 대체 : ML/DL 모델로 결측치 예측

## 11. EDA란? 데이터의 특성 파악(분포, 상관관계)

= EDA. Exploratory Data Analysis 탐색적 데이터 분석은 모델링 전에 데이터의 구조, 분포, 관계, 이상치 등의 특징을 파악하는 과정이다.    (받은 시험지를 살펴보는 과정 이해의 영역)

- 목적
    - 데이터의 전반적인 특성을 이해
    - 노이즈, 이상치, 결측치 탐지
    - 변수 간 관계와 패턴 발견
    - 이후 모델링 방향 및 전처리 전략 결정
- 주요 분석 내용
    - 데이터 분포 : 값들이 어떤 형태로 분포하는지 확인
        - 방법 : 히스토그램, 커널 밀도 추정 등으로
    - 변수 간 상관관계 : 변수들이 가지는 관계를 파악
        - 방법 :  상관계수(두 변수가 함께 움직이는 정도), 시각화, 산점도
    - 데이터 품질 확인
        - 결측치 비율이나 이상치 존재 여부나 범주형 변수의 분포 등

## 12. 회귀에서 절편과 기울기가 의미하는 바는? 딥러닝과 연관되는 것은?  50%

- 졀편(b)
    
    =입력값이 0일 때 출력값의 위치. 즉 직선이 y축과 만나는 지점으로 데이터의 전체 위치를 위아래로 조정하는 역할을 한다.
    
- 기울기(w)
    
    = 입력값이 변할 때 출력값이 얼마나 변하는지를 나타낸다. 즉 독립변수와 종속 변수의 관계 정도와 방향을 표현한다.
    
    -기울기가 크면 입력(독립변수) 변화에 따라 출력(종속변수)가 크케 변화한다.
    
    -기울기가 0이면 입력에 상관없이 출력이 일정하다.
    

딥러닝과의 연관성

-딥러닝은 선형회귀(기울기 + 절편)를 여러 개 쌓아놓은 구조다

-딥러닝의 가장 작은 단위인 뉴런은 사실상 회귀식과 같은 구조를 가진다.

회귀식 여러개를 나열한것과 같은 뉴런의 수식(암기 필요). 즉 그 안에 기울기와 절편도 있다.

회귀 모형의 “기울기, 절편” = 딥러닝의 “가중치, 바이어스”

= 즉 회귀의 기울기는 딥러닝의 가중치 이고 회귀의 절편은 딥러닝의 바이어스

으로 둘 다 “데이터의 추세를 찾는 파라미터”이다.

## 13. 결정트리에서 불순도 - 지니계수란 무엇인가? 역할은?

= 결정트리에서 불순도는 어떤 노드에 섞여 있는 데이터가 얼마나 “혼합되어 있는가(=순수하지 못한가)”를 나타내는 척도.

이 때 많이 쓰이는 지표 중 하나가 지니계수(Gini Index)

- 지니 계수 : 특정 노드에서 어떤 클래스가 나타날 확률.
    - 값이 0이면 한 클래스만 존재(순수), 최대값이 나오면 클래스가 균등하게 섞여 있단 뜻(이진 분류에서는 0.5가 최대값이 됨)
    - 역할
        - 분할 기준 : 결정 트리에서 데이터를 나눌 때 자식 노드들이 부모 노드보다 순수해지도록 나누는 그 지니 계수 감소량을 최대화하는 분할을 찾는다.
        - 분류 품질 평가 : 클래스가 얼마나 잘 분리되었는지 측정
        - 과적합 방지 : 정확도만 보는게 아니라 얼마나 의미있게 잘 나뉘었는지를 보므로 의미 없는 분할을 줄여준다.

## 14. 서포트 벡터머신, 마진, 결정경계란? SVM원리?

- 서포트 벡터 머신(SVM)
    - 분류와 회귀 문제에서 자주 사용되는 지도학습 알고리즘이다.
    - 데이터를 두 클래스 이상으로 나눌 때, 클래스 간의 경계선인 결정 경계를 찾는 것이다. 이 때 클래스 간의 간격을 최대화하는 경계를 찾는 것을 목표로 한다.
- 마진(Margin)
    - 마진은 결정 경계(구분선)와 가장 가까운 데이터 사이의 거리이다.
    - SVM은 이 마진을 최대화하는 결정 경계(구분선)을 찾는다. (군인을 3.8선과 가장 떨어지게 배치)  ⇒ 새로운 데이터가 들어와도 일반화 성능(새로 처음 보는 데이터도 잘 맞춘다)이 좋아진다.
    - 마진이 크면 ? 모델이 데이터를 안정적으로 분류하기 좋다.
- 결정 경계(Decision Boundary)
    - 데이터를 다른 클래스로 구분하는 선,면,초평면으로, 단순 데이터 구분선이 아니라 마진을 최대화하는 최적의 선을 의미한다.
- SVM의 원리
    - 선형 분리 가능한 경우 : 두 클래스가 정확히 구분 된다면, 가능한 모든 직선 중에 마진이 최대가 되는 직선을 찾음.
        - 서포트 벡터 : 그 직선을 만드는 데 영향을 주는 데이터 포인트들
    - 선형 분리 불가능 : 일부 오분류를 허용하면서도 마진을 최대화하도록 조정.
        - Soft Margin(어쩔 수 없이 유도리 있어진 마진)이라고 한다.
    - 비선형 분리 : 현재 차원에서 직선으로 분리가 불가능 할 때, 데이터를 더 높은 차원으로 매핑한다. 비선형 데이터를 고차원 공간으로 변환해 선형 분리가 가능하도록 하는 것!
        - 커널 함수 : 이 매핑(다른 값으로 대입)을 효율적으로 하기 위한 함수

## 15. 앙상블이란?

= 여러개의 다른 모델을 결합하여 하나의 최종 예측 결과를 만드는 방법

-여러 모델을 합침으로 개별 모델의 약점을 보완하고, 과적합을 줄일 수 있음.

- 앙상블 방법
    1. 배깅 : 여러 모델을 병렬로 학습, 최종 예측은 다수결(분류)나 평균(회귀)으로 결정
        1. 대표 알고리즘 : 랜덤 포레스트(Random Forest)
    2. 부스팅 : 여러 모델을 순차적으로 학습, 앞 모델이 틀린 것을 다음 모델이 집중해서 학습, 약한 학습기 순서대로 강화
    3. 스태킹 : 여러 모델의 예측 결과를 메타 모델이 종합하여 최종 예측.

장점 : 일반화 성능이 높아지고 안정성이 높아진다

단점 : 계산량이 증가하고 해석이 어려울 수 있다. ( 여러 모델이 같이 결정하기 때문에 왜 그런 결과가 나왔는지 한 눈에 설명하기 어렵다. )

## 16. 부트 스트랩핑이란?     60%

= 가상데이터셋.  일부를 보고 전체를 맞춰보는 통계적 추정이나 머신러닝 모델의 일반화 성능을 평가하기 위해 데이터를 “복원 추출” 방식으로 무작위 샘플링 하는 기법.

- 방법 : 원래 데이터셋의 크기가 N개이면, 같은 크기로 N개를 무작위 뽑는 것. 복원 추출이기 때문에 같은 데이터가 여러 개 선택 되기도 함. 안 선택 되기도 함.
- 특징
    - 복원추출 : 표본 하나를 뽑고 다시 돌려놓고 다시 뽑기
    - 여러번 반복 : 하나의 원 데이터셋에서 다양한 분포의 샘플들을 얻을 수 있음
    - 통계량 추정(일부를 보고 전체 예측) : 분포와 신뢰구간을 구하는 데 사용
    - 머신러닝 활용
        - 앙상블 기법 활용 : 각 부트스트랩 샘플로 학습한 후 그것으로 일반화 성능 향상
        - 오차 추정 : 샘플링에서 선택되지 않은 데이터들로 모델 성능을 검증
- 장점 :  **부트스트랩 샘플링 → 트리마다 다른 데이터 제공 → 트리들이 다양해짐 → 전체 결과가 더 안정적이고 정확해짐  ex)모의고사 느낌**

## 17. 선형회귀와 로지스틱회귀

= 예측하려는 목표값(타깃 변수)의 성격이 달라서 쓰임새가 다름.

- 선형회귀(Linear Regression)
    - 목적 : 연속적인값을 예측      즉 회귀
    - 출력 : 연속적인 값. 제한이 없는 실수 값
    - 함수 : 단순 선형식
- 로지스틱회귀(Logistic Regression)
    - 목적: 범주(종류)형 값 , 특히 이진 분류를 예측     즉 분류
    - 출력 : 0 ~ 1 사이의 확률값     확률값을 시그모이드 함수에 넣고 0,1을 결정
    - 함수 : 선형식 + 시그모이드 함수를 사용하여 확률이 0.5 이상이면 클래스 1, 아니면 0

분류와 회귀의 기준은 최종 결과에 있다. 계산 방식은 회귀여도 출력이 확률이어도  클래스로 나누면 분류

로지스틱 회귀는 입력값에 선형식을 적용한 결과를 시그모이드 함수에 통과시켜 0~1 사이의 확률값을 출력하고, 이를 임계값 기준으로 클래스를 분류하는 지도학습 알고리즘이다.

## 18. 경사하강법(Gradient Descent)란?

= 머신러닝과 딥러닝 모델이 학습할 때 사용하는 대표적인 최적화 알고리즘으로, 오차(손실 함수, Loss Function)을 줄여가면서 최적의 파라미터인 가중치와 바이어스를 찾아가는 방법이다.

기울기를 따라 조금씩 내려오면서 손실을 최소화하는 방법.

- 손실 함수(L)
    - 모델이 예측한 값과 실제 값의 차이를 수치로 표현한 함수이다.
    - 우리가 줄이고 싶은 대상이다.
- 기울기
    - 손실함수를 가중치에 대하여 미분한 값이다.
    - 기울기가 가르키는 방향은, 오차가 가장 가파르게 증가하는 방향을 뜻한다.
    - 즉 오차를 줄이고 싶으면 기울기의 반대 방향으로 가중치(w)를 조금씩 업데이트 한다.
- 경사하강법의 종류
    1. 배치 경사하강법 : 전체 데이터를 한 번에 계산한다. 안정적이지만 느리다.
    2. 확률적 경사하강법 : 데이터를 한 개씩 사용한다. 불안하지만 빠르다.
    3. 미니배치 경사하강법 : 데이터를 여러 개 묶음으로 사용한다.   속도와 안정성을 둘다 챙기기 때문에 많이 사용.

## 19. 데이터 불균형이 (결과에) 미치는 영향

- 정확도 착시
    - 90개가 정상이고 10개가 비정상인 데이터를 모델이 모두 완벽하다. 라고 평가 했다면? 비정상인 10개를 전혀 찾지 못했지만 그래도 정확도가 90%가 된다.   즉 다수 클래스만 예측해도 정확도가 높게 나오며 겉보기에만 좋아보인다.
- 소수 클래스 무시
    - 드문 클래스를 거의 배우지 못한다. 실제 중요한 결과(질환 판별 등)일 수 있는데 놓칠 가능성이 크다.
- 일반화 성능 저하
    - 훈련 데이터에서는 잘 맞는 것 같아도, 새로운 데이터에서는 소수의 데이터를 더 못 맞추는 경우가 많다.
- 평가 지표 왜곡
    - Accuracy라는 정확도만 보면 성능이 좋아보이지만 Precision(정밀도), Recall(재현율), F1-Score(정확도와 재현율 둘다 동시에 보는 점수) 등을 보면 성능이 낮게 나타난다.    정확도가 다가 아니다!

## 20. 훈련 데이터와 테스트 데이터를 나누는 까닭과 일반화

- 나누는 이유
    - 훈련데이터는 모델이 입력과 정답을 보며 규칙을 학습하는 데 사용하고 테스트 데이터는 모델이 한 번도 보지 못한 새로운 데이터로 모델이 실제로 얼마나 잘 작동하는지를 확인하기 위해 사용한다.
    - 훈련 데이터만 있다면 모델은 단순히 외워서 답을 도출할 것이다. 그래서 모델이 암기가 아니라 일반적인 규칙을 배웠는지 확인하기 위하여 데이터를 나눈다.
- 일반화(Generalization)
    - 모델이 훈련 때 보지 못한 새로운 데이터에 대해서도 좋은 성능을 내는 능력

## 21. 교차 검증이란?

= 데이터를 단순히 훈련, 테스트 데이터로만 나누면 분할 방식에 따라 결과가 달라질 수 있다. 때문에 교차 검증은 데이터를 여러 개의 부분집합(Fold)으로 나누고 번갈아가며 검증에 사용하여 평가의 불공정성을 줄이고 일반화 성능을 더 정확히 측정한다.   대표적인 방식으로 K-Fold가 있다.

- K-Fold Cross Validation(교차 검증)
    1. 전체 데이터를 K개로 균등하게 분할 시킴
    2. K번의 반복에서 (k-1)개의 부분집합은 학습에 사용하고 남은 1개의 부분집합은 검증에 사용함.
    3. 모든 부분집합이 검증에 다 사용된 후 = 즉 K번 반복   평균 성능을 최종 점수로 계산
    - K가 그렇다고 너무 크면 검증 데이터가 작아지며 평가가 불안정해지고 계산량이 증가하는 문제가 있다.

## 22. 평가 지표 종류와 차이점은?

평가지표란 ?

모델이나 알고리즘의 성능을 수치화하여서 비교하고 판단하기 위한 기준이다.

- 회귀 문제에서의 평가 지표 종류
    - MSE : 예측값과 실제값의 차이를 제곱해서 평균한 값   제곱하기 때문에 큰 오차에 민감함
    - RMSE : MSE에 제곱근을 취한 값.   실제 단위와 똑같아서 직관적임
    - MAE : 오차의 절댓갑 평균.   RMSE와 비슷하지만 큰 오차나 작은 오차나 그게 그거임. 균형적임
    - R제곱 : 모델이 데이터를 얼마나 잘 설명하는지를 나타낸다.     상대적 설명력 평가에 적합
- 분류 문제에서의 평가 지표 종류
    - Accuracy정확도 : 전체 샘플 중 맞춘 비율     단 데이터 불균형이 심하면 신뢰도가 떨어짐
    - Precision정밀도 : 양성이라고 예측한 것 중 실제 양성 비율.       전체 기준이 아닌 양성이라 예측한 것만 기준.   틀린양성(False Positive)을 줄임
    - Recall재현율 : 실제 양성 중 모델이 맞춘 비율    진짜 양성 중 얼마나 잡았나.   놓치는 것 줄이기(False Negative를 줄임)
    - F1-score : Precision과 Recall의 조화 평균. 불균형 데이터에서 균형 잡힌 성능 평가를 함.
    - ROC-AUC :  어떤 확률이 true와 false중 어디에 해당하는지를 결정하는 기준을 임계값이라고 하는데, 그 임계값을 0~1로 계속 바꿔보면서 정밀도와 재현율의 성능을 기록하여 만든 곡선 그래프.  1에 가까울수록 성능이 우수한 것.
    - PR-AUC :  특별히 불균형 데이터에서 정밀도와 재현율의 관계를 평가함

분류 문제의 평가 지표는 **혼동행렬** 기반으로 정의됨

혼동행렬이란 ?  모델이 어디서 틀렸는지 자세히 보여주는 도구. 모델이 잘하는 부분과 부족한 부분을 따로 확인할 수 있

True, False는 실제값 종류    Positive, Negative는 예측값 종류

TP : 예측이 양성인데 실제도 양성

FP : 예측이 양성인데 실제는 음성   - 1종 오류

FN : 예측이 음성인데 실제도 음성  -2종 오류

TN : 예측이 음성인데 실제는 양성   - 치명적인 놓침

## 23. Accuracy 말고 F1을 사용하는 이유는?

= 데이터 불균형의 문제 때문(19번 참고)

- 데이터 불균형 상태에서 Accuracy(정확도)는 착시 효과를 줄 수 있음
- F1은 “양성을” 잘 맞추는 능력을 더 공정하게 평가함.   특히 의료 진단, 사기 탐지처럼 소수의 클래스가 더 중요한 경우에 F1이 훨씬 신뢰성 있는 지표가 됨.
- 즉, 정확도는 전체적으로 맞췄는지를 보고 F1은 중요한 소수 클래스를 얼마나 놓치지 않고 잡아냈는지를 봄

## 24. 차원 축소 기법은 무엇인가?

=데이터의 중요한 정보를 최대한 보존하면서, 원래의 고차원 데이터를 더 적은 수의 차원으로 표현하는 기법이다.

차원이 크다 = 수백 수천 개의 특성이 있다. 때문에 축소의 목적은

- 시각화 : 고차원을 2,3차원으로 바꾸면 그래프로 표현 가능해짐
- 노이즈 제거 : 의미 없는 특성은 버리고 중요한 패턴만 남김
- 연산 효율성 향상 : 학습과 예측 속도가 빨라짐
- 과적합 방지 : 불필요한 특성이 줄어들며 모델이 단순해지고 그로인해 일반화 성능이 좋아질 수 있음. (패턴 찾기가 쉬우니까 적용도 쉬워짐)

대표적인 기법

1. 주성분 분석(PCA) : 데이터가 분포되어있는 모양(퍼진 방향)을 찾고 그 방향으로 데이터를 투영(눕힘)하여 차원을 줄인다.
2. 선형판별분석(LDA) : 클랫스 간 분리를 잘 할 수 있는 축을 찾는 방법으로 주로 분류에 활용
3. 비선형 차원 축소 기법 : 어렵
4. 특성 선택 : 기존 특성 중에서 중요한 것만 고르는 방식

약간 시험 범위의 써머리(요약본)을 만드는 느낌인듯.

## 25. KNN, K-means 차이점은?

둘은 완전히 다른 알고리즘.  KNN은 새로운 걸 어디에 분류할지 예측하는 지도학습이고 K-means는 기존 데이터를 비슷한 애들끼리 자동으로 묶는 방법이다.

- KNN(K-Nearest Neighbors)
    - 분류 / 회귀 알고리즘, 지도학습 방식
    - 새로운 데이터가 들어왔을 때 가장 가까운 K개의 이웃 데이터 레이블을 보고 다수결이나 평균으로 예측
    - 핵심은 데이터간의 거리 계산
    - 목적은 예측(분류/회귀)
- K-means
    - 클러스터링 알고리즘, 비지도학습 방식
    - 데이터들을 K개의 그룹으로 나누는 방법: 각 그룹의 가운데를 찾고 각 데이터를 가장 가까운 중심에 할당한다. 중심을 또 재계산하고 이동. 안 바뀔 때까지 계속(=수렴할 때까지 계속)
    - 목적은 군집화

## 27. 머신러닝에서는 왜 데이터 확보가 중요할까?

이유는 크게 3가지    머신러닝 모델은 알고리즘보다 데이터 확보(양+질+다양성)가 더 큰 힘!

1. 학습의 재료가 되기 때문에
    
     : 데이터 기반으로 패턴을 학습하니까 충분한 데이터가 있어야 일반화 능력을 가질 수 있고 훈련 데이터에만 잘 맞춰져서 새 데이터에 대응하지 못하는 과적합 문제를 예방함
    
2. 데이터의 질이 성능을 좌우하기 때문에
    
     : 아무리 알고리즘이 뛰어나도 노이즈가 많거나 잘못된 레이블을 가진 데이터로 학습하면 성능이 떨어짐.
    
3. 편향 문제 방지
    
     : 데이터가 특정 집단이나 상황에 쏠려 있으면 그 편향을 그대로 학습하기 때문. 공정한 AI는 다양하고 균형 잡힌 데이터가 많이 필요함
    
4. 특성 표현력 확보
    
    : 데이터가 많을수록 더 다양한 특징을 학습할 수 있다. 데이터 양뿐만 아니라 데이터의 다양성이 중요하다.
    
5. 튜닝 및 검증 가능성
    
    : 데이터가 많아야 훈련/검증/테스트를 나누어 평가할 수 있음. 부족하면 제대로 성능 검증하기 어려움
    

## 28. 인공지능에서 왜 Bias는 해결할 수 없는 난제인가?

- 데이터 자체가 이미 여러 사회적 기술적 환경의 영향을 받아 불완전하거나 불균형하기 때문에 완벽히 공정하게 만드는 것은 불가함.
- Biass를 제거하려면 “공정성”의 기준이 필요한데 그 정읜ㄴ 모호함. 완벽히 만족하는 보편적인 공평성 기준이 존재하지 않음
- 모델이 복잡하고 또 편향의 원인에 대해 볼 수 없음.(비가시성) 편향이 어디서 커지는지 명확히 알기 어려움
- 설령 현재 데이터에서 Biass를 줄여도 새로운 데이터가 들어오면 또 다른 형태의 편향이 생김
    
    즉 Biass(편향)은 정적 문제가 아니라 동적 문제. 사라지지 않음.
    
- 편향을 줄이려다가 성능이 저하되거나 다른 집단에서의 불평등이 심화될 수 있음

= 편향은 줄이고 관리하는 것이 목표이지, 없애는 것은 불가능함.

## 29. 손실함수와 Gradient 관계

- 손실 함수 : 모델이 얼마나 잘 못 예측했는지를 수치로 표현함.
- 손실함수를 미분 → 가중치 w가 손실에 얼마나 영향을 주는지 기울기를 알 수 있음 → 기울기의 크기와 방향을 통해 가중치를 얼마나 어느 방향으로 조정해야 손실이 적어지는지 알 수 있음 → 학습률을 곱해 실제 업데이트 크기를 정함.
- 기울기 = 얼마나 조정할지에 대한 조정 비율의 방향과 상대적 크기
- 학습률 = 한 번에 얼마만큼 이동할지에 대한 절대적 크기를 조절

## 30. Activation function(활성화 함수)를 사용하는 이유는?

=신경망이 단순한 선형 계산을 넘어 복잡한 비선형 패턴을 학습하고 다양한 문제에 맞게 출력을 조정할 수 있게 하기 위함이다.

크게 3가지 이유

1. 비선형성 부여 : 신경망이 단순히 선형으로만 이루어져 있다면 아무리 층을 많이 쌓아도  결국 하나의 선형 모델이 됨. 따라서 활성화 함수를 통해 비선형성을 넣어주어야 신경망이 복잡한 패턴을 학습할 수 있음.
2. 복잡한 함수 근사 : 입력과 출력의 관계가 매우 복잡해도 신경망이 그 함수를 학습할 수 있게 함.
3. 출력 범위 조정 : 일부 활성화 함수는 출력값을 0 ~ 1 사이로 제한(시그모이드)하거나 -1 ~ 1 사이로 제한(탄)하여 해석이나 스케일 조정에 유리하게 함.  ReLU 등은 음수는 0으로 만들고 양수만 전달해서 **희소성**과 학습효율을 높임(많은 0중에 별로 없는 실수값이 생기는 거니까)

ex) Sigmoid(시그모이드), Tanh(탄), ReLU, Softmax 여러 클래스 중 하나일 확률 분포로 변환     등이 대표적인 예시 함수임.

=신경망에서 각 층의 출력에 비선형성을 부여하여 복잡한 데이터의 패턴을 학습할 수 있도록 하는 함수이다. 활성화 함수가 없을 시에 신경망은 계속해서 선형적인 관계만을 학습하여 아무리 층을 더 쌓아도 결국 다른 것이 없는 선형 모델이 된다. 때문에 비선형적인 관계를 학습할 수 있도록 하는 활성화 함수가 필요하다.

## 31. Softmax 함수란?

= 출력값의 어떤 수를 확률처럼 변환하여 상대적인 크기 차이를 확률 형태로 변환하는 함수. 이 함수를 통해 가장 가능성 높은 걸 뚜렷하게 확인할 수 있어서 다중분류 문제에 주로 사용한다.

즉 여러 점수 ⇒ 각 점수가 정답일 확률 로 바꿈, 점수를 **확률처럼** 해석하게 함.

왜 확률이 아니라 확률처럼 일까?  

: 모델이 어떤 답을 주장하는 자신감의 점수(정도)를 숫자로 나타내어 그걸 확률처럼 보이게 만든 것이지 실제 확률처럼 그 답을 보장할 수는 없음. 실제 확률은 90%에 대해 정말 90%이지만. 소프트맥스함수의 확률은 90%라고 생각해! 느낌? 모델의 예측이 항상 정확하지 않기 때문에 그 값을 그대로 확률로 해석하기는 힘들다.

즉, 모델의 자신감 점수를 우리가 보기 편하게 확률의 형태로 확률처럼 만든 것.  

- AI 답변
    
    여러 클래스에 대한 점수(score)를 입력받아 각 값을 0~1 사이로 정규화하고 전체 합이 1이 되도록 하여 확률처럼 해석할 수 있는 값으로 변환하는 함수이다. 이를 통해 다중 분류 문제에서 가장 가능성이 높은 클래스를 선택할 수 있다.
    

## 32. Gradient Vanishing / Exploding 문제란?

- Gradient Vanishing 기울기 소실이란
    - 정의 : 역전파 과정에서 기울기가 점점 작아져 결국 0에 수렴하는 문제 (전달 놀이를 하는데 점점 약하게 때리면 마지막엔 느낌도 안 나게 됨)
    - 원인 : 활성화 함수이 미분값이 0에 가까운 구간이 많음 → 곱셈 반복 시에 기울기가 급격히 작아짐, 네트워크가 깊을수록 이 상황이 누적됨.
    - 결과 : 초기 층의 가중치 업데이트가 거의 일어나지 않음. 학습 속도가 느려지며 ⇒ 심층 신경망 학습이 불가능해짐
- Gradient Exploding 기울기 폭발이란
    - 정의 : 역전파 과정에서 기울기가 층을 거치며 점점 커져 무한히 발산하는 문제
    - 원인 : 가중치 값이 너무 크거나 적절히 정규화 되지 않았을 때 RNN 등의 시퀀스 모델에서 긴 시퀀스를 다루며 곱셈이 누적되어 값이 커짐
    - 결과 : 가중치 값이 비정상적으로 커지며 학습이 불안정해짐 ⇒ 손실 함수가 NaN(Not a Number)으로 발산됨.(숫자로 계산 불가, 계산이 깨짐)
- 해결 방법
    - 소실 완화
        - ReLU 계열 활성화 함수 사용 : 음수 구간에서도 기울기가 0이 되지 않게 유지
        - 적절한 가중치 초기화 기법
        - Batch Normalization 적용
        - ResNet 구조 도입 : 입력을 출력에 더해줘서 기울기가 깊은 층까지 직접 전달되게 함
    - 폭발 완화
        - Gradient Clipping : 임계값 이상은 잘라내기
        - 적절한 학습률 설정 : 너무 큰 업데이트로 인해 값이 급격히 커지는 것을 방지하기
        - 정규화 기법 : 가중치의 크기를 제한해서 모델이 과도하게 커지는 것을 방지하기
