### 1. CNN
이미지의 공간적 정보를 유지하며 특징을 추출하는 신경망. 돋보기로 이미지를 흝음

#### 핵심개념
Feature Extraction : 필터(커널)를 사용하여 이미지의 국소적 특징을 계산  <br>
Parameter Sharing : 필터를 공유함으로 덴스레이어 대비 가중치 수를 확 줄이고 효율성을 높임<br>
Stride & Padding : 연산 범위와 출력 크기를 조절하며 경계 손실을 방지함 <br>
Pooling (Max/Average) : 주요 특징은 보존하고 데이터 크기를 줄여 연산 효율성을 높이며, 미세한 위치 변화에 강인한 모델을 만든다.<br><br>
- Stride : 돋보기를 몇 칸씩 옮길 것인가
- Padding : 이미지 가장자리가 잘리지 않게 테두리를 더 둘러주는 것
- Pooling : 너무 세세한 정보까지는 버리고 중요한 특징만 남겨서 용량을 아끼는 것. Max Pooling은 가장 큰 값만 남기는 것


Danse Layer(밀집층)<br>
: 입력되는 모든 데이터가 출력되는 모든 노드오 하나도 빠짐없이 연결된 상태<br>
모두가 모두를 연결! 입력 데이터가 3개고 출력 노드가 2개라면? 총 6개의 연결선(가중치)가 생긴다.

### 2. RNN과 LSTM
시계열 데이터나 자연어처럼 순서가 중요한 데이터를 처리하는 신경망

#### RNN
: 다음에 올 말을 맞추듯, 순서가 있는 데이터를 처리함. 이전의 정보를 기억해서 다음 단계로 전달! 이전 단계의 출력이 현재 단계의 입력에 영향을 주는 루프 구조

#### RSTM
: RNN의 단점.. 좀 큰 단점인 '금방 까먹음' 건망증 문제를 해결한 모델.
Forget Gate, Input Gate, Output Gate를 통해 장기 기억과 단기 기억을 관리하여 긴 문장과 같은 데이터를 효과적으로 처리.
 중요한 정보는 더 오래 기억하고 필요 없는 것은 지우는 장치가 추가되어 있음.
